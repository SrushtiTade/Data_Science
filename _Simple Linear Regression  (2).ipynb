# Independent And Dependent Variable

```python
x=df.drop('Salary',axis=1)
```

```python
x.head()
```

```python
y=df.Salary
```

```python
y.head()
```

```python
y.head(10)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=0,test_size=0.30)
```

```python
print(x_train.shape)
```

```python
x_test.shape
```

# Model FITTING 

```python
from sklearn.linear_model import LinearRegression
LR=LinearRegression()
LR.fit(x_train,y_train)
```

```python
#Assignment coefficient (slope) to m
m=LR.coef_
```

```python
print("coefficient:",m)
```

```python
#Assigning y-imtercept to a
c=LR.intercept_
```

```python
print("Intercept :",c)
```

# Evaluation MEtrics

```python
from sklearn import metrics
```

```python
Accuracy = LR.score(x_test,y_test)
Accuracy
```
